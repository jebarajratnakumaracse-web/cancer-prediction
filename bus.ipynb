```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# Create figure and axis
fig, ax = plt.subplots(figsize=(10,5))

# Bus body
body = patches.Rectangle((1,2), 8, 2, edgecolor='black', facecolor='blue')
ax.add_patch(body)

# Bus top
top = patches.Rectangle((2,4), 5, 1, edgecolor='black', facecolor='skyblue')
ax.add_patch(top)

# Windows
for i in range(5):
    window = patches.Rectangle((1.5 + i*1.2, 3), 0.8, 0.7,
                               edgecolor='black', facecolor='lightyellow')
    ax.add_patch(window)

# Door
door = patches.Rectangle((7.5,2), 1, 1.5,
                         edgecolor='black', facecolor='brown')
ax.add_patch(door)

# Wheels
wheel1 = patches.Circle((2.5,1.8), 0.5, color='black')
wheel2 = patches.Circle((7,1.8), 0.5, color='black')

ax.add_patch(wheel1)
ax.add_patch(wheel2)

# Road line
ax.plot([0,10],[1.3,1.3])

# Settings
ax.set_xlim(0,10)
ax.set_ylim(1,6)
ax.set_aspect('equal')
plt.axis('off')

# Show bus
plt.show()
```
